```python
#size attributes

import numpy as np
arr = np.arange(1,15,2)
print(arr)

sz = arr.size
print("Array size: ",sz)
```

```python
#shape attributes

import numpy as np

arr = np.arange(16).reshape(4,4)
print(arr)

sp = arr.shape
print("The shape of the array is : ",sp)
```

```python
#ndim attributes

import numpy as np
arr = np.arange(16).reshape(4,4)
print(arr)

dim = arr.ndim
print("The dimension of the array is : ",dim)
```

```python
#nbytes attributes

import numpy as np
arr = np.arange(16).reshape(4,4)
print(arr)

bt = arr.nbytes
print("The number of bytes in the array is : ",bt)
```

```python
#dtype attributes

import numpy as np
arr = np.arange(16).reshape(4,4)
print(arr)

dt = arr.dtype
print("The datatype of the array is : ",dt)
```
